#Exercise 1: Campus Resource Tracker
#Problem 1.
A university wants to track the availability of resources such as projectors, laptops, and laboratory equipment. Develop a program that maintains resource information and helps users identify
available resources.
Each resource should contain a unique ID, a name, a category, and the available quantity. You may represent
each resource using a list or use a list of dictionaries if dictionaries have already been covered.
Sample data
1 resources = [
2
[101, "Laptop", "Electronics", 8],
3
[102, "Projector", "Electronics", 0],
4
[103, "Whiteboard", "Classroom", 5],
5
[104, "Microphone", "Audio", 3],
6
[105, "Camera", "Electronics", 2]
7 ]
Tasks
1. Write a function display_resources(resources) to display all resources with their details.
2. Write a function search_by_category(resources, category) to display all resources belonging to a
given category.
3. Write a function find_unavailable(resources) to identify resources whose available quantity is zero.
Testing and Debugging
Test your program with an unknown resource ID, a negative quantity, a resource with zero units, and an
empty resource list. Explain how your program handles each case.

```python

resources = [
    [101, "Laptop", "Electronics", 8],
    [102, "Projector", "Electronics", 0],
    [103, "Whiteboard", "Classroom", 5],
    [104, "Microphone", "Audio", 3],
    [105, "Camera", "Electronics", 2]
]


def display_resources(resources):
    if len(resources) == 0:
        print("No resources available.")
        return

    for resource in resources:
        print("ID:", resource[0])
        print("Name:", resource[1])
        print("Category:", resource[2])
        print("Available Quantity:", resource[3])


def search_by_category(resources, category):
    found = False

    for resource in resources:
        if resource[2].lower() == category.lower():
            print(resource)
            found = True

    if found == False:
        print("No resource found in this category.")


def find_unavailable(resources):
    found = False

    for resource in resources:
        if resource[3] == 0:
            print(resource[1], "is unavailable.")
            found = True

    if found == False:
        print("All resources have available units.")


# Display all resources
print("All Resources:")
display_resources(resources)

# Search by category
print("\nElectronics Resources:")
search_by_category(resources, "Electronics")

# Find unavailable resources
print("\nUnavailable Resources:")
find_unavailable(resources)
```

Develop a program that processes student records and reports useful information about their results.
Each student record contains a student ID, a name, and marks from three assessments. Marks must be between
0 and 100.

Tasks
1. Write a function calculate_average(marks) that returns the average of a student’s marks.
2. Write a function classify_performance(average) that returns a performance category using these rules:
• Average ≥ 80: Excellent
• Average ≥ 70: Good
• Average ≥ 60: Satisfactory
• Average < 60: Needs Improvement
3. Write a function display_report(students) that displays each student’s ID, name, average, and perfor
mance category.
Testing and Debugging
Test your program with a student who has equal marks in all assessments, students with tied averages, an
invalid mark, and a student record with no marks. Decide how invalid or missing data should be handled and
implement the appropriate checks.

```python

students = [
    [201, "Amina", [78, 85, 90]],
    [202, "Rahim", [65, 70, 68]],
    [203, "Nadia", [92, 88, 95]],
    [204, "Hasan", [55, 60, 65]],
    [205, "Tania", [80, 80, 80]]
]


def calculate_average(marks):
    if len(marks) == 0:
        return -1

    total = 0

    for mark in marks:
        if mark < 0 or mark > 100:
            return -1

        total = total + mark

    average = total / len(marks)
    return average


def classify_performance(average):
    if average >= 80:
        return "Excellent"
    elif average >= 70:
        return "Good"
    elif average >= 60:
        return "Satisfactory"
    else:
        return "Needs Improvement"


def display_report(students):
    if len(students) == 0:
        print("No student records available.")
        return

    for student in students:
        student_id = student[0]
        name = student[1]
        marks = student[2]

        average = calculate_average(marks)

        if average == -1:
            print("ID:", student_id)
            print("Name:", name)
            print("Invalid marks or no marks available.")

        else:
            performance = classify_performance(average)

            print("ID:", student_id)
            print("Name:", name)
            print("Average:", round(average, 2))
            print("Performance:", performance)


display_report(students)
```
